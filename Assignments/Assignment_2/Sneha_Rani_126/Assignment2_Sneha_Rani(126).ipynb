# Ques 1 Find Sum of First 10 Natural Numbers

```python
sum = 0
for i in range(1, 11): #range is used 11 not included
    sum= sum + i

print("Sum of first 10 natural numbers =", sum)
```

# Ques 2 Find Factorial of a Number

```python
num = int(input("Enter a number: "))

fact = 1
for i in range(1, num + 1):
    fact= fact*i

print("Factorial =", fact)
```

# Ques 3 Fibonacci Series

```python
n=int(input("enter how many terms in series= "))
def fibonacci(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b #next term will be sum of prevoius two numbers

fibonacci(n)
```

# Ques 5 Student Result System

```python
# Student Result System

# Input student details
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

# Input marks
mark1 = float(input("Enter marks of Subject 1: "))
mark2 = float(input("Enter marks of Subject 2: "))
mark3 = float(input("Enter marks of Subject 3: "))

# Calculate total and percentage
total = mark1 + mark2 + mark3
percentage = total / 3

# Display grade
if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

# Display result
print("\n----- STUDENT RESULT ----\n")
print("Name       :", name)
print("Roll No    :", roll_no)
print("Total Marks:", total)
print("Percentage :", percentage)
print("Grade      :", grade)
```
